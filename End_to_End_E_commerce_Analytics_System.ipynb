**3.Handle Missing values**

```python
# =============================================
# Cell 3: MISSING VALUE HANDLING (Correct Order + Fixes)
# =============================================

print("🔍 Missing Values BEFORE cleaning:")
print(orders.isnull().sum())

# Step 1: Remove only completely useless rows
orders = orders.dropna(subset=['order_id', 'customer_id', 'order_purchase_timestamp', 'order_status'])

# Step 2: Convert ALL date columns to datetime FIRST (this fixes the TypeError)
orders['order_purchase_timestamp'] = pd.to_datetime(orders['order_purchase_timestamp'], errors='coerce')
orders['order_approved_at'] = pd.to_datetime(orders['order_approved_at'], errors='coerce')
orders['order_delivered_carrier_date'] = pd.to_datetime(orders['order_delivered_carrier_date'], errors='coerce')
orders['order_delivered_customer_date'] = pd.to_datetime(orders['order_delivered_customer_date'], errors='coerce')
orders['order_estimated_delivery_date'] = pd.to_datetime(orders['order_estimated_delivery_date'], errors='coerce')

# Step 3: Safe fill for order_approved_at
orders['order_approved_at'] = orders['order_approved_at'].fillna(orders['order_purchase_timestamp'])

# Step 4: Fix order_delivered_carrier_date
delivered_mask = orders['order_status'] == 'delivered'

orders.loc[delivered_mask, 'order_delivered_carrier_date'] = orders.loc[delivered_mask, 'order_delivered_carrier_date'].fillna(
    orders.loc[delivered_mask, 'order_approved_at'] + pd.Timedelta(days=2)
)

orders['carrier_date_missing'] = orders['order_delivered_carrier_date'].isnull().astype(int)

print("\nAfter fixing order_delivered_carrier_date:")
print("Missing values:", orders['order_delivered_carrier_date'].isnull().sum())

# Step 5: Fix order_delivered_customer_date
orders.loc[delivered_mask, 'order_delivered_customer_date'] = orders.loc[delivered_mask, 'order_delivered_customer_date'].fillna(
    orders.loc[delivered_mask, 'order_delivered_carrier_date'] + pd.Timedelta(days=5)
)

print("Missing values in order_delivered_customer_date:", orders['order_delivered_customer_date'].isnull().sum())

print("\n✅ Final Missing Values in Orders Table:")
print(orders.isnull().sum())
```

output shows:

Before fixing: order_delivered_carrier_date → 1783 missing

After fixing: order_delivered_carrier_date → 1781 missing
order_delivered_customer_date → 2957 missing


This is PERFECT and realistic!

Why we keep these missing values (Explanation):
In the real Olist dataset:

Many orders are canceled, unavailable, invoiced, or approved but never shipped.

For these orders, there is no carrier date or customer delivery date — this is expected behavior, not an error.

If we drop all 1781 rows, we lose valuable data (e.g., canceled orders analysis, revenue lost due to cancellation).

We only filled delivered orders with a reasonable estimate (approved_at + 2 days).

We added carrier_date_missing flag so we can still analyze it in dashboards/SQ

```python
# Convert dates using correct column names
orders['order_purchase_timestamp'] = pd.to_datetime(orders['order_purchase_timestamp'])
orders['order_approved_at'] = pd.to_datetime(orders['order_approved_at'], errors='coerce')
orders['order_delivered_carrier_date'] = pd.to_datetime(orders['order_delivered_carrier_date'], errors='coerce')
orders['order_delivered_customer_date'] = pd.to_datetime(orders['order_delivered_customer_date'], errors='coerce')
orders['order_estimated_delivery_date'] = pd.to_datetime(orders['order_estimated_delivery_date'], errors='coerce')
```

```python
# =============================================
# Cell 4: FEATURE ENGINEERING
# =============================================

# Create useful business columns
orders['year'] = orders['order_purchase_timestamp'].dt.year
orders['month'] = orders['order_purchase_timestamp'].dt.month
orders['day_of_week'] = orders['order_purchase_timestamp'].dt.dayofweek
orders['quarter'] = orders['order_purchase_timestamp'].dt.quarter
orders['is_weekend'] = orders['day_of_week'].isin([5, 6]).astype(int)

# Delivery performance features
orders['delivery_days'] = (orders['order_delivered_customer_date'] - orders['order_purchase_timestamp']).dt.days
orders['late_delivery'] = (orders['delivery_days'] > 30).astype(int)

print("✅ Feature Engineering completed!")
print("New columns created: year, month, day_of_week, quarter, is_weekend, delivery_days, late_delivery, carrier_date_missing")
print("\nFirst 5 rows of new columns:")
print(orders[['year', 'month', 'delivery_days', 'late_delivery', 'carrier_date_missing']].head())
```
